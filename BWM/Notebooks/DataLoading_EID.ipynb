```python
# Create a directory storing responses for all eids.
import os
import numpy as np
import pandas as pd
from tqdm import tqdm
import sys

# Add project root to path to import bwm_functions
sys.path.append(os.path.abspath('..'))
from bwm_functions import get_eid_response_matrix
```

```python
directory_path = '../Data2'
output_directory = '../ResponseData/EID_PseudoResponses'

if not os.path.exists(output_directory):
    os.makedirs(output_directory)

# Get list of EIDs
all_items = os.listdir(directory_path)
eids = [item for item in all_items if os.path.isdir(os.path.join(directory_path, item))]
print(f"Found {len(eids)} EIDs in {directory_path}")

```

```python
# Parameters for response matrix
event = "stimOn_times"
conditions = ['choice']
n_steps = 100
n_pseudo = 25
threshold = 1.0 # Minimum mean firing rate
```

```python
for eid in tqdm(eids, desc="Processing EIDs"):
    output_file = os.path.join(output_directory, f"{eid}_pseudo_responses.npz")
    
    # Skip if already processed
    if os.path.exists(output_file):
        continue
        
    result = get_eid_response_matrix(
        eid, directory_path, event, conditions, 
        n_steps=n_steps, n_pseudo=n_pseudo, threshold=threshold
    )
    
    if result is not None:
        res_matrix, cond_matrix, beryl_regions, cosmos_regions = result
        
        # Save as npz to include metadata
        np.savez_compressed(
            output_file,
            res_matrix=res_matrix,
            cond_matrix=cond_matrix,
            beryl_regions=beryl_regions,
            cosmos_regions=cosmos_regions
        )
```
