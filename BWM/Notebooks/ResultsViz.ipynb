```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
import numpy as np

# Create a sample DataFrame (using numerical variables for simplicity)
data = pd.DataFrame({
    'X1': [1, 2, 3, 4, 5],
    'X2': [2, 3, 4, 5, 6],
    'y': [4, 7, 10, 13, 16]
})

# Define the features and target
X = data[['X1', 'X2']]
y = data['y']

# Use PolynomialFeatures to generate interaction terms (degree=2 includes main effects and all 2-way interactions)
# include_bias=False prevents adding an extra intercept term, as LinearRegression() handles the intercept by default
poly = PolynomialFeatures(degree=2, interaction_only=True, include_bias=False)
X_interact = poly.fit_transform(X)

# The new feature names
print("Features names:", poly.get_feature_names_out())

# Fit the linear regression model with interaction terms
model = LinearRegression()
model.fit(X_interact, y)

# Print the coefficients
print("Coefficients:", model.coef_)
```

```python
neuron_indices = df.sort_values('pca_error').neurons.values[0]
eid = "35ed605c-1a1a-47b1-86ff-2b56144f55af"
directory_path = "Data"
time_bin = .01
# Construct the full path to the randomly selected folder
full_path_to_random_folder = os.path.join(directory_path, eid)
print(f"Randomly selected eid: {full_path_to_random_folder}")
spike_count = load_npz(full_path_to_random_folder + '/responses.npz').toarray()


print("reshaping spikes")
spike_count = spike_count.reshape(-1,100,spike_count.shape[1])

#
print("Smoothing")
trial_responses = np.zeros(spike_count.shape)
for neuron in range(spike_count.shape[2]):
    for trial in range(spike_count.shape[0]):
        trial_responses[trial,:,neuron] = gaussian_filter1d(spike_count[trial,:,neuron]/time_bin,sigma = 5)

# Restriction to active units
responses = trial_responses[:,:,np.std(trial_responses,axis=(0,1))>.1]
responses = responses[:,:,np.mean(responses,axis=(0,1))>.1]


device = 'cpu'
y = torch.tensor(responses).float().to(device)
# Subsample neurons
#neuron_indices = np.random.choice(y.shape[2], 100,replace = False)
y = y[:,:,neuron_indices]

# Z-score single neuron responses
y = (y - torch.mean(y, dim=[0, 1], keepdim=True)) / torch.std(y, dim=[0, 1], keepdim=True)


train_indices = np.random.choice(len(y), int(.8 * y.shape[0]), replace=False)
test_indices = list(set(range(len(y))) - set(train_indices))
y_train = y[train_indices]
y_test = y[test_indices]

print("Initializing RNN")
nonlinear_model = RNN(n=y.shape[2], input_dim=32, latent_dim=64, history_len=100,activation = nn.ReLU()).to(
    device)
nonlinear_model.fit(steps=y.shape[1], epochs=2000,batch_size = 100, lr=.001, verbose=True, weight_decay=0., y_train=y_train,
          y_test=y_test,
)
```
